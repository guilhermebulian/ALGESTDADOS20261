```python
#03 - Sistema Bancário de Saques
# Um sistema bancário simples precisa simular operações de saque em uma conta corrente. O programa deve iniciar com um saldo informado pelo usuário e permitir saques sucessivos até que seja digitado 0. A cada tentativa, verifique se há saldo suficiente, atualize o saldo quando possível e informe ao usuário quando o valor exceder o saldo disponível.

saldo = float(input('Insira o saldo da conta: R$'))
saque = float(input('Insira o valor do saque (0 para encerrar): R$'))

while (saque != 0):

  if (saque <= saldo):
    saldo = saldo - saque
    print(f'Saque realizado com sucesso.')
    print(f'Saldo atual: R${saldo:.2f}')

  else:
    print('Saldo insuficiente.')

  saque = float(input('\nInsira o valor do saque (0 para encerrar): R$'))

print(f'\nSaldo final: R${saldo:.2f}')
```

```python
#28 - Caça aos Primos. Um sistema deve analisar números inteiros informados pelo usuário até que seja digitado -1. Ao final, informe quantos números são primos e calcule a soma dos números que não são primos.
contP = 0
somnP = 0

while True:
    N = int(input('Digite um número inteiro (-1 para finalizar): '))
    if N == -1:
        break
    P = False
    if N <= 1:
        P = False
    else:
        totD = 0
        for i in range(1, N + 1):
            if N % i == 0:
                totD = totD + 1
        if totD == 2:
            P = True
        else:
            P = False
    if P == True:
        contP = contP + 1
    else:
        somnP = somnP + N
print('\n--- Resultados da Caça aos Primos ---')
print(f'Quantidade de números primos: {contP}.')
print(f'Soma dos números que não são primos: {somnP}.')
```

```python
#29 - Avaliação de Notas
# Um sistema de avaliação deve ler notas até que seja digitado 0, calcular a média geral e informar quantas notas estão acima da média.

soma = 0
quantidade = 0
acimaMedia = 0
nota = float(input('Digite a nota (0 para encerrar): '))

while (nota != 0):

  soma = soma + nota
  quantidade = quantidade + 1
  nota = float(input('Digite a nota (0 para encerrar): '))

media = soma / quantidade
nota = float(input('\nDigite novamente as notas para verificar as acima da média (0 para encerrar): '))

while (nota != 0):

  if (nota > media):
    acimaMedia = acimaMedia + 1
  nota = float(input('Digite novamente as notas para verificar as acima da média (0 para encerrar): '))

print(f'\nMédia geral: {media:.2f}')
print(f'Notas acima da média: {acimaMedia}')
```

```python
#30 - Cálculo de Fatorial. Um programa deve ler um número inteiro positivo e calcular seu fatorial utilizando repetição. Após isso, deve verificar se o resultado do fatorial é par ou ímpar e apresentar essa informação ao usuário.
Nint = 0

while True:
    Nint = int(input('Digite um número inteiro positivo para calcular o fatorial: '))
    if Nint > 0:
        break
    else:
        print('Por favor, digite um número inteiro POSITIVO.')
F = 1
for i in range(1, Nint + 1):
    F = F * i
print(f'\nO fatorial de {Nint} é: {F}.')
paRimp = ""
if F % 2 == 0:
    paRimp = "PAR"
else:
    paRimp = "ÍMPAR"
print(f'O resultado ({F}) é um número {paRimp}.')
```
