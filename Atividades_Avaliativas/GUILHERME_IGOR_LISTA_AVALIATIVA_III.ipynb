```python
#02 Escreva um programa que leia 10 nomes e 10 idades. Armazene os nomes em uma lista e as idades em outra. Em seguida, exiba o nome da pessoa mais velha.

nomes = []
idades = []
print("Por favor, digite 10 nomes e idades:")
for i in range(10):
    nome = input(f"Digite o nome da {i+1}ª pessoa: ")
    nomes.append(nome)
    while True:
        idade = input(f"Digite a idade de {nome}: ")
        intV = True
        TempIdad = idade
        if not TempIdad:
            intV = False
        else:
            inicio = 0
            if TempIdad[0] == '-':
                inicio = 1
                if len(TempIdad) == 1:
                    intV = False
            if intV:
                for carac in TempIdad[inicio:]:
                    if not ('0' <= carac <= '9'):
                        intV = False
                        break
        if intV:
            idade = int(TempIdad)
            idades.append(idade)
            break
        else:
            print("Entrada inválida. Por favor, digite um número inteiro para a idade.")
if idades:
    maxIda = max(idades)
    imv = idades.index(maxIda)
    nmv = nomes[imv]
    print(f"\nA pessoa mais velha é: {nmv} com {maxIda} anos.")
else:
    print("Nenhuma idade foi inserida.")
```

```python
#08 Faça um programa que simule o sorteio de 6 números da Mega-Sena (entre 1 e 60) sem repetição. Armazene em uma lista e exiba.

MODULO = 61
MULTIPLICADOR = 23
INCREMENTO = 17
atual = 1
numS = []
while len(numS) < 6:
    atual = (MULTIPLICADOR * atual + INCREMENTO) % MODULO
    if atual == 0:
        numero_gerado = 60
    else:
        numero_gerado = atual
    repetido = False
    for num_existente in numS:
        if numero_gerado == num_existente:
            repetido = True
            break
    if not repetido:
        numS.append(numero_gerado)
tamanho_lista = len(numS)
for i in range(tamanho_lista):
    for j in range(0, tamanho_lista - i - 1):
        if numS[j] > numS[j+1]:
            temp = numS[j]
            numS[j] = numS[j+1]
            numS[j+1] = temp
print("Números sorteados da Mega-Sena:", numS)
```

```python
#17 - Dada uma matriz 3×3 com valores reais, calcule o determinante (sem usar bibliotecas externas). Use apenas listas e estruturas de repetição.
matriz = []

for linha in range(3):
    lista = []
    for coluna in range(3):
        valor = float(input(f'Digite o valor da posição [{linha}][{coluna}]: '))
        lista.append(valor)
    matriz.append(lista)

a = matriz[0][0]
b = matriz[0][1]
c = matriz[0][2]
d = matriz[1][0]
e = matriz[1][1]
f = matriz[1][2]
g = matriz[2][0]
h = matriz[2][1]
i = matriz[2][2]

determinante = (a * e * i) + (b * f * g) + (c * d * h) - (c * e * g) - (a * f * h) - (b * d * i)
print(f'Determinante:{determinante}')
```

```python
#18 Faça um programa que leia duas matrizes 3×3 e calcule a matriz soma e a matriz diferença, armazenando os resultados em novas listas.

dim = 3
mat1 = []
print("\nDigite os elementos da primeira matriz 3x3:")
for i in range(dim):
    lin1 = []
    for j in range(dim):
        while True:
            try:
                val = int(input(f"Elemento da 1ª matriz [{i+1}][{j+1}]: "))
                lin1.append(val)
                break
            except:
                print("Entrada inválida. Digite um número inteiro.")
    mat1.append(lin1)
mat2 = []
print("\nDigite os elementos da segunda matriz 3x3:")
for i in range(dim):
    lin2 = []
    for j in range(dim):
        while True:
            try:
                val = int(input(f"Elemento da 2ª matriz [{i+1}][{j+1}]: "))
                lin2.append(val)
                break
            except:
                print("Entrada inválida. Digite um número inteiro.")
    mat2.append(lin2)
matSom = [[mat1[i][j] + mat2[i][j] for j in range(dim)] for i in range(dim)]
matDif = [[mat1[i][j] - mat2[i][j] for j in range(dim)] for i in range(dim)]
print("\nMatriz 1:")
for lin in mat1:
    print(lin)
print("\nMatriz 2:")
for lin in mat2:
    print(lin)
print("\nMatriz Soma:")
for lin in matSom:
    print(lin)
print("\nMatriz Diferença:")
for lin in matDif:
    print(lin)
```

```python
#19 - Solicite dois conjuntos de 10 elementos fornecidos pelo usuário. Mostre a interseção, união e diferença (A − B e B − A).
conjunto_a = set()
conjunto_b = set()

print('Digite os 10 elementos do conjunto A:')
for i in range(10):
    numero = int(input())
    conjunto_a.add(numero)

print('Digite os 10 elementos do conjunto B:')
for i in range(10):
    numero = int(input())
    conjunto_b.add(numero)

intersecao = conjunto_a & conjunto_b
uniao = conjunto_a | conjunto_b
diferenca_ab = conjunto_a - conjunto_b
diferenca_ba = conjunto_b - conjunto_a

print(f'Interseção: {intersecao}')
print(f'União: {uniao}')
print(f'Diferença A - B: {diferenca_ab}')
print(f'Diferença B - A: {diferenca_ba}')
```

```python
#29 - Crie um dicionário onde as chaves são nomes de produtos e os valores são listas com preços praticados em três lojas diferentes. Encontre o produto com maior variação de preço.

produtos = {}
for i in range(5):

    nome = input('Digite o nome do produto: ')
    precos = []

    for j in range(3):
        preco = float(input(f'Digite o preço da loja {j + 1}: '))
        precos.append(preco)
    produtos[nome] = precos

maior_variacao = 0
produto_maior_variacao = ''

for produto in produtos:

    precos = produtos[produto]
    maior_preco = precos[0]
    menor_preco = precos[0]

    for preco in precos:

        if preco > maior_preco:
            maior_preco = preco
        if preco < menor_preco:
            menor_preco = preco

    variacao = maior_preco - menor_preco
    if variacao > maior_variacao:
        maior_variacao = variacao
        produto_maior_variacao = produto

print(f'Produto com maior variação de preço: {produto_maior_variacao}')
print(f'Variação: {maior_variacao}')
```
